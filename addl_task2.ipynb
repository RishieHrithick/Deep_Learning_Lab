```python
import matplotlib.pyplot as plt

class Perceptron:

    def __init__(self,lr=1):

        self.lr=lr
        self.w=[0,0]
        self.b=0

    def predict(self,x):

        z=self.w[0]*x[0]+self.w[1]*x[1]+self.b

        if z>=0:
            return 1
        else:
            return 0

    def train(self,X,Y):

        update=1

        for epoch in range(15):

            print("\nEpoch",epoch+1)

            errors=0

            for x,y in zip(X,Y):

                pred=self.predict(x)

                diff=y-pred

                if diff!=0:

                    self.w[0]+=self.lr*diff*x[0]
                    self.w[1]+=self.lr*diff*x[1]
                    self.b+=self.lr*diff

                    print("Update",update)
                    print("Weights =",self.w)
                    print("Bias =",self.b)

                    self.plot(X,Y,update)

                    update+=1
                    errors+=1

            print("Errors =",errors)

        print("\nTraining stopped because convergence was not achieved.")

    def plot(self,X,Y,step):

        plt.figure(figsize=(5,5))

        for x,y in zip(X,Y):

            if y==0:
                plt.scatter(x[0],x[1],color='red',s=100)

            else:
                plt.scatter(x[0],x[1],color='blue',s=100)

        if self.w[1]!=0:

            xs=[-0.5,1.5]

            ys=[]

            for x in xs:
                ys.append((-self.w[0]*x-self.b)/self.w[1])

            plt.plot(xs,ys)

        plt.xlim(-0.5,1.5)
        plt.ylim(-0.5,1.5)

        plt.title("XOR Update "+str(step))

        plt.grid(True)

        plt.show()


X=[
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]

Y=[0,1,1,0]

p=Perceptron()

p.train(X,Y)
```
