```python
import matplotlib.pyplot as plt

# -------------------------
# Perceptron Class
# -------------------------

class Perceptron:

    def __init__(self, lr=1):
        self.lr = lr
        self.w = [0,0]
        self.b = 0

    def predict(self,x):

        z = self.w[0]*x[0] + self.w[1]*x[1] + self.b

        if z >= 0:
            return 1
        else:
            return 0

    def train(self,X,Y,title):

        update = 1

        for epoch in range(20):

            error = 0

            print("\nEpoch",epoch+1)

            for x,y in zip(X,Y):

                pred = self.predict(x)

                diff = y-pred

                if diff!=0:

                    self.w[0]+=self.lr*diff*x[0]
                    self.w[1]+=self.lr*diff*x[1]
                    self.b+=self.lr*diff

                    print(f"Update {update}")
                    print("Weights =",self.w)
                    print("Bias =",self.b)

                    self.plot_boundary(X,Y,title,update)

                    update+=1
                    error+=1

            if error==0:
                print("\nTraining Converged")
                break


    def plot_boundary(self,X,Y,title,step):

        plt.figure(figsize=(5,5))

        for x,y in zip(X,Y):

            if y==0:
                plt.scatter(x[0],x[1],color='red',s=100)

            else:
                plt.scatter(x[0],x[1],color='blue',s=100)

        x_values=[-0.5,1.5]

        if self.w[1]!=0:

            y_values=[]

            for x in x_values:
                y=(-self.w[0]*x-self.b)/self.w[1]
                y_values.append(y)

            plt.plot(x_values,y_values)

        plt.xlim(-0.5,1.5)
        plt.ylim(-0.5,1.5)

        plt.title(title+" Update "+str(step))

        plt.grid(True)
        plt.show()
```

```python
X = [
    [0,0],
    [0,1],
    [1,0],
    [1,1]
]

Y = [0,1,1,1]

p = Perceptron(lr=1)

p.train(X,Y,"OR Gate")
```
